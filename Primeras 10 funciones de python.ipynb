DESARROLLO: 30 FUNCIONES BÁSICAS DE PYTHON

1. Función print()

Descripción:

La función print() permite mostrar información en la pantalla. Se utiliza para presentar mensajes, resultados de operaciones y valores almacenados en variables.

Ejemplo:

El siguiente código muestra el nombre y la edad de una persona.

```python
Nombre = "Fernanda"

edad = 26

print("Mi ombre es:", Nombre)

print("Mi edad es:", edad)
```

2. Función len()

Descripción:

La función len() permite conocer la cantidad de elementos que contiene una colección o la cantidad de caracteres que tiene una cadena de texto.

Ejemplo práctico:

El siguiente programa cuenta cuántas letras tiene el nombre Alejandro.

```python
nombre = "Alejandro"

cantidad = len(nombre)

print("El nombre es:", nombre)
print("Cantidad de letras:", cantidad)
```

3. Función type()

Descripción:

La función type() permite identificar el tipo de dato que contiene una variable. Por ejemplo, permite saber si un valor es un número entero, un número decimal o una cadena de texto.

Ejemplo práctico:

El siguiente programa identifica los tipos de datos de tres variables que contienen el nombre, la edad y la estatura de una persona.

```python
nombre = "Dilan"
edad = 25
estatura = 1.85

print("El tipo de nombre es:", type(nombre))
print("El tipo de edad es:", type(edad))
print("El tipo de estatura es:", type(estatura))
```

4. Función int()

Descripción: La función int() permite convertir un número o un dato compatible en un número entero, es decir, sin decimales.

Ejemplo: Convertir un número decimal en un número entero.

```python
numero = 25.8

resultado = int(numero)

print("Número original:", numero)
print("Número convertido a entero:", resultado)
```

5. Función float()

Descripción: La función float() permite convertir un número entero o un dato compatible en un número decimal. Es útil cuando necesitamos trabajar con cantidades que tienen decimales.

Ejemplo: Convertir un número entero en un número decimal.

```python
numero = 58

resultado = float(numero)

print("Número original:", numero)
print("Número convertido a decimal:", resultado)
```

6. Función str()

Descripción: La función str() permite convertir un dato en una cadena de texto. Es útil cuando necesitamos unir números con mensajes o mostrar información como texto.

Ejemplo: Convertir un número entero en texto.

```python
edad = 48

resultado = str(edad)

print("Edad original:", edad)
print("Edad convertida a texto:", resultado)
print("Tipo de dato:", type(resultado))
```

7. Función bool()

Descripción: La función bool() permite convertir un dato en un valor booleano, es decir, en True (verdadero) o False (falso). Se utiliza para comprobar si un valor se considera verdadero o falso dentro de un programa.

Ejemplo práctico: Comprobar si una persona tiene una edad mayor que cero.

```python
edad = 25

resultado = bool(edad > 0)

print("Edad:", edad)
print("¿La edad es mayor que cero?:", resultado)
```

8. Función abs()

Descripción: La función abs() permite obtener el valor absoluto de un número. Esto significa que convierte los números negativos en positivos, mientras que los números positivos permanecen iguales.

Ejemplo práctico: Calcular la distancia entre dos valores sin tener en cuenta el signo negativo.

```python
numero = -250

resultado = abs(numero)

print("Número original:", numero)

print("Valor absoluto:", resultado)
```

9. Función round()

Descripción: La función round() permite redondear un número decimal al número entero más cercano o a la cantidad de decimales que se indique. Se utiliza para simplificar valores numéricos.

Ejemplo práctico: Redondear el promedio de una calificación a dos cifras decimales.

```python
promedio = 58.8597

resultado = round(promedio, 2)

print("Promedio original:", promedio)
print("Promedio redondeado:", resultado)
```

10. Función max()

Descripción: La función max() permite identificar y devolver el valor más grande dentro de un conjunto de números o elementos. Se utiliza para comparar datos y encontrar el valor máximo.

Ejemplo práctico: Identificar la calificación más alta de un grupo de estudiantes.

```python
calificaciones = [150.10, 15.2, 175.8, 89.92, 41.5]

mayor = max(calificaciones)

print("Calificaciones:", calificaciones)
print("La calificación más alta es:", mayor)
```
